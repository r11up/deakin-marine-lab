```python
from PIL import Image

# Constants
data_path = "/content/drive/MyDrive/Capstone/Deakin/furseal/reduced_masked_dataset_128x128/"
image_size = 128  # 128x128
max_dim = 2**7    # 128 features (7 qubits)

X, y = [], []

# Create label map A=0, B=1, ..., H=7
label_map = {folder: idx for idx, folder in enumerate(sorted(os.listdir(data_path)))}

for folder, label in label_map.items():
    folder_path = os.path.join(data_path, folder)
    for file in os.listdir(folder_path):
        img = Image.open(os.path.join(folder_path, file)).convert("L").resize((image_size, image_size))
        vec = np.array(img).flatten().astype(np.float32)
        vec = vec / np.linalg.norm(vec)  # amplitude encoding requires norm-1
        vec = vec[:max_dim]              # trim to 128 dims for 7 qubits
        X.append(vec)
        y.append(label)

X = np.array(X)
y = np.array(y)
print(f"Shape of encoded data: {X.shape}, Labels: {np.unique(y)}")
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.25, random_state=42)
```

```python
n_qubits = 7  # Because 2^7 = 128

def amplitude_feature_map(x):
    qc = QuantumCircuit(n_qubits)
    qc.append(StatePreparation(x), qc.qubits)
    return qc
```

from qiskit.primitives import Sampler


```python
# Use Qiskit primitives
sampler = Sampler()
fidelity = Fidelity(sampler=sampler)

# Define fidelity kernel with amplitude encoding feature map
kernel = FidelityQuantumKernel(feature_map=amplitude_feature_map, fidelity=fidelity)

# Pegasos QSVC (Quantum SVM)
pegasos = PegasosQSVC(quantum_kernel=kernel, C=1.0)

# Train model
pegasos.fit(X_train, y_train)
```

```python
# Predict and score
y_pred = pegasos.predict(X_test)
acc = accuracy_score(y_test, y_pred)

print(f"PegasosQSVC accuracy with amplitude encoding: {acc:.4f}")
```
